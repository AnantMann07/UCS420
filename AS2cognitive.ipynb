```python
roll = "1024170112"

L = [int(i) * 10 for i in roll]

print("THE MULTIPLIED LIST:",L)

L.append(90)
print("APPEND:",L)

L.insert(2, 50)
print("INSERT:",L)

L.remove(90)
print("REMOVE:",L)

L.pop(2)
print("POP:",L)

L.sort()
print("SORT ASEC:",L)

L.sort(reverse=True)
print("SORTING DESC",L)

print("FIRST THREE:",L[:3])
print("LAST THREE",L[-3:])

avg = sum(L) / len(L)
new_list = [x for x in L if x > avg]

print(new_list)
```

```python
scores = tuple(L[:8])

print(scores)

highest = max(scores)
print("HIGHEST",highest)
print(scores.index(highest))

lowest = min(scores)
print("LOWEST",lowest)
print(scores.count(lowest))

reverse_scores = list(scores[::-1])
print(reverse_scores)

x = int(input("Enter a score: "))

if x in scores:
    print(scores.index(x))
else:
    print("Not present")



first, second, *remaining = scores

print("FIRST",first)
print("SECOND",second)
print("REMAINING",remaining)

scores[0] = 100
```

```python
import random

random.seed(1024170112)

numbers = [random.randint(100, 900) for i in range(100)]

odd = [x for x in numbers if x % 2 != 0]
even = [x for x in numbers if x % 2 == 0]

print("Odd numbers:", len(odd))
print("Even numbers:", len(even))

def prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

primes = [x for x in numbers if prime(x)]

print("Prime numbers:", len(primes))
print(primes)

most = max(numbers, key=numbers.count)

print("Most frequent:", most)
print("Count:", numbers.count(most))
```
