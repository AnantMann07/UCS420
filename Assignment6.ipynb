```python
#ques1 a
import math

p = 0.2
q = 0.8
k = 6
x = 25

prob = math.comb(x + k - 1, k - 1) * (p ** k) * (q ** x)

print("Probability =", prob)
```

```python
#ques1 b
import math

p = 0.2
q = 0.8
k = 6

prob_less_20 = 0

for x in range(20):
    prob_less_20 += math.comb(x + k - 1, k - 1) * (p ** k) * (q ** x)

prob_at_least_20 = 1 - prob_less_20

print("Probability =", prob_at_least_20)
```

```python
#ques1 c
import numpy as np
import matplotlib.pyplot as plt

p = 0.2
k = 6

results = np.random.negative_binomial(k, p, 1000)

plt.hist(results, bins=20)
plt.xlabel("Number of Failures")
plt.ylabel("Frequency")
plt.title("Negative Binomial Simulation")
plt.show()
```

```python
#ques2a
p = 0.13
q = 0.87

x = 8

prob = (q ** x) * p

print("Probability =", prob)
```

```python
#ques 2b
p = 0.13
q = 0.87

prob = 0

for x in range(6, 11):
    prob += (q ** x) * p

print("Probability =", prob)
```

```python
#ques 2c
import numpy as np
import matplotlib.pyplot as plt

p = 0.13

results = np.random.geometric(p, 200) - 1

plt.hist(results, bins=15)
plt.xlabel("Number of Calls")
plt.ylabel("Frequency")
plt.title("Seatbelt Survey Simulation")
plt.show()
```

```python
#ques4a
import math

p = 0.55
q = 0.45

prob = math.comb(5, 3) * (p ** 4) * (q ** 2)

print("Probability =", prob)
```

```python
#ques4b
import math

p = 0.55
q = 0.45

prob_4 = p ** 4

prob_5 = math.comb(4, 3) * (p ** 4) * q

prob_6 = math.comb(5, 3) * (p ** 4) * (q ** 2)

prob_7 = math.comb(6, 3) * (p ** 4) * (q ** 3)

total = prob_4 + prob_5 + prob_6 + prob_7

print("Probability =", total)
```

```python
#ques4c
import math

p = 0.55
q = 0.45

prob_3 = p ** 3

prob_4 = math.comb(3, 2) * (p ** 3) * q

prob_5 = math.comb(4, 2) * (p ** 3) * (q ** 2)

total = prob_3 + prob_4 + prob_5

print("Probability =", total)
```
