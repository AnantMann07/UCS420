```python
#ques3
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 5
p = 0.42

expected = n * p
print("Expected heads =", expected)

sd = np.sqrt(n * p * (1-p))
print("Standard deviation =", sd)

x = np.arange(0, 6)
prob = binom.pmf(x, n, p)

plt.bar(x, prob)
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.title("Probability Distribution")
plt.show()
```

```python
#example 2
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Given values
n = 10
p = 0.5

# x = 0 to 10
x = np.arange(0, 11)

# Probability mass function
prob = binom.pmf(x, n, p)

# Print x and probabilities
print("x =", x)
print("Probability =", prob)

# Probability bar plot
plt.bar(x, prob)
plt.xlabel("x")
plt.ylabel("P(X = x)")
plt.title("Binomial Probability Distribution")
plt.show()

# Cumulative probability
cumulative = binom.cdf(x, n, p)

# Print cumulative probabilities
print("Cumulative Probability =", cumulative)

# Cumulative probability graph
plt.plot(x, cumulative, marker='o')
plt.xlabel("x")
plt.ylabel("P(X <= x)")
plt.title("Cumulative Probability Distribution")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Given
lam = 6

# (a) Probability of exactly 4 calls
exactly_4 = poisson.pmf(4, lam)
print("P(X = 4) =", exactly_4)

# Probability of at least 4 calls
at_least_4 = 1 - poisson.cdf(3, lam)
print("P(X >= 4) =", at_least_4)


# (b) Probability histogram
x = np.arange(0, 20)
prob = poisson.pmf(x, lam)

plt.bar(x, prob)
plt.xlabel("Number of Calls")
plt.ylabel("Probability")
plt.title("Poisson Distribution of Calls")
plt.show()


# (c) Simulate 50 periods
simulation = np.random.poisson(lam, 50)

print("50 Simulated Values:")
print(simulation)
```

```python
#exercise 1
import numpy as np
from scipy.stats import binom

# Given
p = 1/4

# (a) Fires 7 times
n = 7

# At least 2 hits = 1 - P(X <= 1)
answer_a = 1 - binom.cdf(1, n, p)

print("P(X >= 2) =", answer_a)


# (b) Find minimum number of fires
for n in range(1, 100):
    probability = 1 - (1-p)**n

    if probability > 2/3:
        print("Required number of fires =", n)
        print("Probability =", probability)
        break
```

```python
#exercise 2
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Given
n = 10
p = 0.7

# (a) Exactly 8 shots
answer_a = binom.pmf(8, n, p)
print("P(X = 8) =", answer_a)

# (b) At most 6 shots
answer_b = binom.cdf(6, n, p)
print("P(X <= 6) =", answer_b)

# (c) Simulate 50 rounds
simulation = np.random.binomial(n, p, 50)

print("50 Simulated Rounds:")
print(simulation)

# Frequency distribution
values, frequency = np.unique(simulation, return_counts=True)

plt.bar(values, frequency)
plt.xlabel("Number of Successful Shots")
plt.ylabel("Frequency")
plt.title("Frequency Distribution of Successes")
plt.show()
```
