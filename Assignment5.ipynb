```python
# Question 2
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

print("Original Array:", arr)
print("Reversed Array:", arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print()

print("Most frequent value:", most_frequent)
print("Indices:", indices)
print()

y = np.array([1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
# Question 4
import numpy as np

Anantpreet = np.linspace(10, 100, 25)

print("Array:")
print(Anantpreet)

print("Dimensions:", Anantpreet.ndim)
print("Shape:", Anantpreet.shape)
print("Total elements:", Anantpreet.size)
print("Data type:", Anantpreet.dtype)
print("Total bytes:", Anantpreet.nbytes)


transposed = Anantpreet.reshape(1, 25)

print("Transpose using reshape:")
print(transposed)

# Transpose using T
print("Transpose using T:")
print(Anantpreet.T)
```

```python
# Question 5
import numpy as np


ucs420_Anantpreet = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Anantpreet)
print()


print("Mean:", np.mean(ucs420_Anantpreet))


print("Median:", np.median(ucs420_Anantpreet))


print("Maximum:", np.max(ucs420_Anantpreet))


print("Minimum:", np.min(ucs420_Anantpreet))


print("Unique elements:", np.unique(ucs420_Anantpreet))


ucs420_Anantpreet = ucs420_Anantpreet.reshape(4, 3)

print("\nAfter Reshaping to 4 x 3:")
print(ucs420_Anantpreet)

# Resize to 2 rows and 3 columns
resized_ucs420_Anantpreet = np.resize(ucs420_Anantpreet, (2, 3))

print("\nAfter Resizing to 2 x 3:")
print(resized_ucs420_Anantpreet)
```
